Otra forma de hacerlo es usar los `defaultdict` de la librerias `collections`. La cual nos permite asinar un valor default basado en la definición de un tipo:

| # | type            | syntax                    | default value |
|---|-----------------|---------------------------|---------------|
| 1 | int             | defaultdict(int)          | 0             |
| 2 | str             | defaultdict(str)          | ''            |
| 3 | list            | defaultdict(list)         | []            |
| 4 | dict            | defaultdict(dict)         | {}            |
| 5 | lambda function | defaultdict(lambda x: ….) | lambda(x)     |

````python
from collections import defaultdict

fruits = defaultdict(int)
fruits['coco'] = 1000
fruits['orange'] = 1200
fruits['pineaple'] = 1200

print(fruits)
````

````python
fruits["strawberry"] # valore default es cero
````

````python
authors = defaultdict(list)
authors['dostoievsky'] = ["crimen y castigo", "hermano karamazov"]
authors['shakespeare'] = ["otelo", "macbeth", "el rey Lear"]

authors["dickens"]
````
