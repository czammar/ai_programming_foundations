Además se puede combinar lo anterior rebanar una sublista elegindo:

## 3. Métodos para Listas 

| Method    | Description                                                                  |
|-----------|------------------------------------------------------------------------------|
| append()  | Adds an element at the end of the list                                       |
| clear()   | Removes all the elements from the list                                       |
| copy()    | Returns a copy of the list                                                   |
| extend()  | Add the elements of a list (or any iterable), to the end of the current list |
| index()   | Returns the index of the first element with the specified value              |
| insert()  | Adds an element at the specified position                                    |
| pop()     | Removes the element at the specified position                                |
| remove()  | Removes the first item with the specified value                              |
| reverse() | Reverses the order of the list                                               |
| sort()    | Sorts the list                                                               |

````python
characters.append("bowser")
````

````python
characters
````

````python
len(characters) # Ahora hay 6 personajes!!
````

````python
# Agreguemos mas personajes
characters.append("daisy")
characters.append("boo")
characters.append("bowser")
````

````python
characters # Bowser aparece varias veces!
````

````python
characters.index("bowser") # Lugar donde "bowser" aparece por primera vez
````

````python
characters.pop(5) # Quita el elemento en posicion 5
````

````python
characters
````

````python
# Ahora volvemos a buscarlo
characters.index("bowser")
````

````python
characters.insert(6, "koopa troopa") # Agregamos un nuevo personaje en la posición 6
````

````python
characters
````

La lista también se puede ordenar con `sort()` y `reverse()`.
**Nota:** Ambos metodos afectan permanente a la lista original!!!

# 4. For Loop & Comprehension Lists

Las listas se pueden recorrer con for loops

````python
for numero in magic_numbers:
    print(numero)
````

Del mismo modo, se pueden usan para crear listas nuevas a partir de otra, como por ejemplo
```
[i for i in my_list]
```

````python
# Cuadrados de enteros del 0 al 10
[e**2 for e in range(10)]
````

````python
# ¿Qué es lo que calcula el siguiente código?
# Hint: Usa el comando help(sum)
#sum([e**2 for e in range(100000)])
````

**Q:** ¿Qué hace el operador `in` en el siguiente código?

````python
# Personajes cuyo nombre contiene la letra "i"
[personaje for personaje in characters if ("i" in personaje)]
````
