B. El cifrado Ceaser es una técnico de critografía muy antigua que parte de las siguientes ideas:
* a) Considerar un conjunto ordenado de simbolos como alfabeto, por ejemplo `ABCDEFGHIJKLMNOPQRSTUVWXYZ`,
* b) Para cifrar un mensaje, cada letra se desplaza una cantidad de lugares. En la siguiente figura se muestra cómo se desplazarian las letras del alfabeto usando 2 espacios:

![](./images/ceaser_cipher.png)

La siguiente función hace el cifrado de mensajes en mayusculas y minusculas:

```python
SYMBOLS = 'ABCDEFGHIJKLMNOPQRSTUVWXYZabcdefghijklmnopqrstuvwxyz'

def caesar_cypher(message: str, displacement: int = 2) -> str:
    """
    Crea un cifrado cesar de una palabra desplazandose cierta cantidad
    de caracteres.
    """
    translated = ''
    key = displacement

    for symbol in message:
        if symbol in SYMBOLS:
            symbol_index = SYMBOLS.find(symbol)
            translated_index = (symbol_index + key) % len(SYMBOLS)
            if translated_index < 0:
                translated_index = translated_index + len(SYMBOLS)
            translated = translated + SYMBOLS[translated_index]
        else:
            translated = translated + symbol

    return translated
```

* FIFO: First In - First Out


![queue](https://cdn1.byjus.com/wp-content/uploads/2021/11/queue-dequeue.png)

```python
messages = ["message_" + str(i) for i in range(20) ]
```

```python
messages
```

```python
from collections import deque

q = deque()

for msg in messages:
    q.append(msg)
```

```python
q.popleft()
```

```python
q.popleft()
```

```python
q.popleft()
```
