# Modelo Lineal de Juguete

$$ \text{median\_house\_value} = constant + coef_1*\text{longitude} + coef_2*\text{latitude} \\
+ coef_3*\text{housing\_median\_age} + coef_4*\text{total\_rooms} + \\
coef_5*\text{total\_bedrooms} + coef_6*\text{population} + \\
coef_7*\text{households} + coef_8*\text{median\_income} 
$$

```python
# Inicialización de coeficientes con valor 1
constant               = 0
coef_longitude          = 1
coef_latitude           = 1
coef_housing_median_age = 1
coef_total_rooms        = 1
coef_total_bedrooms     = 1
coef_population         = 1
coef_households         = 1
coef_median_income      = 1
```

```python
prediccion = (
    constant +
    (coef_longitude          * house_example[0]) +
    (coef_latitude           * house_example[1]) +
    (coef_housing_median_age * house_example[2]) +
    (coef_total_rooms        * house_example[3]) +
    (coef_total_bedrooms     * house_example[4]) +
    (coef_population         * house_example[5]) +
    (coef_households         * house_example[6]) +
    (coef_median_income      * house_example[8])  # Saltando el índice 7
)
```

```python
print("prediccion:", prediccion)
```

```python
error = abs(prediccion-house_example[8])
```
