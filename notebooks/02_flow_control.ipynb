# 1. While Loop

Es un bloque de codigo que se ejecuta hasta que una condicion es cierta

```
while my_condition:
    # my magic code 
    # my_condition being recalculated
```

````python
m = 1

while m < 15:
    m = m +1
    print(m)

print("hola ya salimos de while")
````

**Q:** ¿cómo hago una funcion que sume los cubos de los numeros de 1 al N, donde N es un parámetro?

Queremos una funcion que te ponga un nombre en mayusculas y lo repita N veces

````python
batman="bruce wayne"
batman.upper()*5
````

````python
def process_name(name, n):
    # lo pasa a mayusculas
    name_capitalized = name.upper()
    # lo repite n veces
    name_repetead = name_capitalized*n
    return name_repetead
````

````python
process_name("pikachu poKeMOn", 3)
````
