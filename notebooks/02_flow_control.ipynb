# 1. While Loop

Es un bloque de codigo que se ejecuta hasta que una condicion es cierta

```
while my_condition:
    # my magic code 
    # my_condition being recalculated
```

````python
m = 1

while m < 6:
    m = m +1
    print(m)
````

# 3. If ... Else

````python
price_car =  50000

if price_car > 100000:
    print("price_car is very expensive")
else:
    print("price_car is cheap")
````

````python
price_car =  36000

if price_car < 35000:
    print("price_car is a bargain")
elif price_car < 80000:
    print("price_car is kind of expensive")
else:
    print("price_car is cheap")
````
