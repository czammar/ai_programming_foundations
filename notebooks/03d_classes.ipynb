```python
class Dog:
    """
    Clase que modela a un perro
    """
    # Constructor de atributos
    def __init__(self, name, age):
        # Datos que la clase va a guardar
        self.name = name
        self.age = age

    # Metodos
    def bark(self):
        return f"{self.name} te esta ladrando!"

    def age_msg(self):
        return f"El perro tiene {self.age} años"
```

### Cars

```python
class Automovil:
    """
    Clase que modela un automóvil con capacidad de acelerar y frenar.
    """
    def __init__(self, marca: str, modelo: str, velocidad_actual: float = 0.0):
        self.marca = marca
        self.modelo = modelo
        self.velocidad_actual = velocidad_actual
        self.status = self.determina_status()

    def info(self):
        """Obtiene informacion del auto."""
        return f"El auto {self.marca} {self.modelo} va a {self.velocidad_actual} km/h."

    def acelerar(self, incremento: int, turbo: bool = False) -> str:
        """
        Incrementa la velocidad_actual del automóvil por un valor.
        Modifica el atributo self.velocidad_actual.
        """
        if turbo==True:
            incremento=incremento*10
        self.velocidad_actual = self.velocidad_actual + incremento
        return f"Acelerando. Velocidad actual: {self.velocidad_actual} km/h."

    def determina_status(self) -> str:
        """
        Ayuda a determinar si el auto esta detenio o en movimiento
        """
        if self.velocidad_actual > 0:
            status = "en movimiento"
        else:
            status = "detenido"
        return status
```

```python
auto1 = Automovil(marca="BMW", modelo="X1", velocidad_actual=0)
```

```python
auto1.info()
```

```python
auto1.determina_status()
```

```python
auto1.acelerar(incremento=5)
```

```python
class Customer:
    def __init__(self, customer_id, name, email):
        self.customer_id = customer_id
        self.name = name
        self.email = email
        self.purchase_history = [] # Lista vacia

    def update_email(self, new_email):
        "Funcion para actualizar el email"
        self.email = new_email

    def add_purchase(self, item, amount, date):
        self.purchase_history.append({'item': item, 'amount': amount, 'date': date})

    def get_purchase_history(self):
        return self.purchase_history
```

# 2. Dataclasses 

Como vimos, las clases de Python se pueden usar para guardar y procesar datos en Python:

```python
class Flower:
    def __init__( self, sepal_length, sepal_width, petal_length, petal_width, species):
        self.sepal_length = sepal_length
        self.sepal_width = sepal_width
        self.petal_length = petal_length
        self.petal_width = petal_width
        self.species= species
```

```python
flor1 = Flower(5.1, 3.5, 1.4, 0.2, "Setosa")
```

```python
print(flor1)
```

```python
flor1.sepal_length
```

```python
flor1.species
```

Una alternativa es usar dataclasses: https://docs.python.org/3/library/dataclasses.html

```python
from dataclasses import dataclass

@dataclass # Decorador
class Flower:
    # Se definen los campos con anotaciones de tipo
    sepal_length: float
    sepal_width: float
    petal_length: float
    petal_width: float
    species: str
```

```python
flower_data = Flower(5.1, 3.5, 1.4, 0.2, "Setosa")
```

```python
print(flower_data)
```

# 3. Herencia

```python
class Figura:
    """
    Clase para representa
    """
    def __init__(self, nombre: str, largo: float, ancho: float):
        self.nombre = nombre
        self.largo = largo
        self.ancho = ancho

    def info(self):
        "mensaje con informacion general"
        return f"{self.nombre}, con largo: {self.largo} y ancho {self.ancho}"
```

```python
class Rectangulo(Figura):
    """
    Clase para representa
    """
    def __init__(self, nombre: str, largo: float, ancho: float, fecha):
        # Le pasamos los argumentos a la clase padre para definirlo
        super().__init__(nombre, largo, ancho)
        # La clase puede tener sus propios argumentos
        self.fecha = fecha

    def calcula_area(self):
        "Metodo nuevo que no tiene la clase original"
        return self.largo*self.ancho
    
    def calcula_perimetro(self):
        "Metodo nuevo que no tiene la clase original"
        return 2*(self.largo+self.ancho)
```

```python
rec = Rectangulo(nombre="rectangulo", largo=5, ancho=12, fecha="03-11-2025")
```

```python
rec.info()
```

```python
rec.calcula_area()
```

```python
rec.calcula_perimetro()
```

```python
rec.fecha
```

```python
class Person:
    def __init__(self, name):
        self.name = name

    def get_details(self):
        return f"Esta persona se llama {self.name}"

class Athlete:
    def __init__(self, sport):
        self.sport = sport

    def get_skill(self):
        return f"Esta persona practica {self.sport}"

class Student(Person, Athlete):
    def __init__(self, name, sport, major):
        # Person recibe el valor 'name' para definir self.name
        Person.__init__(self, name)

        # Athlete recibe el valor sport para definir self.sport
        Athlete.__init__(self, sport)

        # Valor self.major
        self.major = major

    def get_major(self):
        return f"Esta persona estudia {self.major}"

# Example usage
student = Student(name="Wednesday", sport="Fencing", major="Literature")
print(student.get_details())
print(student.get_skill())
print(student.get_major())
```

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split

iris = load_iris()
```

```python
iris.data[:10]
```

```python
iris.target[10:]
```

```python
X = iris.data
y = iris.target
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
```

```python
from sklearn.tree import DecisionTreeClassifier, plot_tree
import matplotlib.pyplot as plt
```

```python
model = DecisionTreeClassifier(max_depth=2, max_features=2)
```

```python
model.__dict__
```

```python
model.fit(X_train, y_train)
```

```python
plot_tree(model, filled=True)
plt.title("Decision tree trained on all the iris features")
plt.show()
```

```python
y_test_pred = model.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score
```

```python
print(f"Accuracy: {accuracy_score(y_test, y_test_pred):.2f}")
```
