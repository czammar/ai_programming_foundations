# Learning Curves for Cross Validation

```python
import matplotlib.pyplot as plt
from sklearn.datasets import make_classification
from sklearn.model_selection import learning_curve
from sklearn.ensemble import RandomForestClassifier

# Datos de ejemplo
X, y = make_classification(n_samples=2500, n_features=200, random_state=42)

# Curva de aprendizaje
train_sizes, train_scores, val_scores = learning_curve(
    RandomForestClassifier(n_estimators=10, max_depth=5), X, y, cv=10, scoring='accuracy'
)
```

```python
# Graficar
plt.plot(train_sizes, train_scores.mean(axis=1), label='Entrenamiento')
plt.plot(train_sizes, val_scores.mean(axis=1), label='Validación')
plt.xlabel('Tamaño del conjunto de entrenamiento')
plt.ylabel('Precisión')
plt.title('Curva de Aprendizaje')
plt.legend()
plt.show()
```
