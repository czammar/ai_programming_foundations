**Fermat Little Theorem**

Sean $a$ y $p$ dos números enteros, tales que $p$ es primo. Entonces los siguientes valores son iguales:

* El residuo de $a^p$ dividido entre $p$, y
* El residuo de $a$ dividido entre $p$

Comprúebalo:

```python
a = 2007
p = 3
```

```python
# el residuo de a dividido entre p
a % p
```

```python
# el residuo de a la p dividido entre p
a**p % p
```

```python
print(p)
```

| Operador | Significado                           | Ejemplo              | Resultado | Descripción breve |
|-----------|----------------------------------------|----------------------|------------|--------------------|
| `>`       | Mayor que                              | `5 > 3`              | `True`     | Comprueba si el valor de la izquierda es mayor. |
| `<`       | Menor que                              | `2 < 8`              | `True`     | Comprueba si el valor de la izquierda es menor. |
| `=`       | Asignación                             | `x = 10`             | —          | Asigna un valor a una variable (no es comparación). |
| `==`      | Igual a                                | `5 == 5`             | `True`     | Comprueba si los valores son iguales. |
| `!=`      | Distinto de                            | `4 != 5`             | `True`     | Comprueba si los valores son diferentes. |
| `>=`      | Mayor o igual que                      | `7 >= 7`             | `True`     | Comprueba si el valor de la izquierda es mayor o igual. |
| `<=`      | Menor o igual que                      | `3 <= 5`             | `True`     | Comprueba si el valor de la izquierda es menor o igual. |



```python
a = True
b = False

print("a > b:", a > b)
print("a < b:", a < b)
print("a == b", a == b)
print("a != b:", a != b)
print("a >= b:", a >= b)
print("a <= b:", a <= b)
```
