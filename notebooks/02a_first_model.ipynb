```python
def model_dummy(sample):
    """
    Clasifica una flor del dataset Iris según las reglas
    de un árbol de decisión.
    Parámetro:
        sample: tupla con formato (
        sepal_length, sepal_width, petal_length, petal_width, species
        )
    Retorna:
        str: clase predicha ('Setosa', 'Versicolor', 'Virginica', o '2')
    """
    
    # Extraer características relevantes
    petal_length = sample[2]
    petal_width = sample[3]
    
    # Criterio para predecir
    if petal_length > 5:
        specie="Setosa"
    elif petal_width < 1:
        specie="Versicolor"
    else:
        specie="Virginica"
    return specie
```

```python
def model_tree(sample):
    """
    Clasifica una flor del dataset Iris según las reglas
    de un árbol de decisión.
    Parámetro:
        sample: tupla con formato (
        sepal_length, sepal_width, petal_length, petal_width, species
        )
    Retorna:
        str: clase predicha ('Setosa', 'Versicolor', 'Virginica', o '2')
    """
    
    # Extraer características relevantes
    petal_length = sample[2]
    petal_width = sample[3]
    
    # Árbol de decisión basado en reglas
    if petal_length <= 2.45:
        specie="Setosa"
    else:
        if petal_width <= 1.75:
            if petal_length <= 4.95:
                specie="Versicolor"
            else:
                specie="Virginica"
        else:
            if petal_length <= 4.85:
                specie="Virginica"
            else:
                specie="Virginica"
    return specie
```

```python
aciertos = 0
for flower in flowers:
    specie = flower[4]
    if model_tree(flowers[130]) == specie:
        aciertos = aciertos +1
```
