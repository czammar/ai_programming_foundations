```python
df["Date"] = pd.to_datetime(df["Date"])
```

```python
df["year"] = df["Date"].dt.year
df["month"] = df["Date"].dt.month
df["day"] = df["Date"].dt.day
```

```python
# figura y eje de la figura
fig, ax = plt.subplots(figsize=(10, 5))

# numero de intervalos para conteos
n_bins = 100

# creacion del objeto historgama
n, bins, patches = ax.hist(
    df['Price'],
    n_bins
    )

ax.set_title('Fig.1 Histograma de frequencias de Price')
ax.set_xlabel('Price de las casas')
ax.set_ylabel('# Ocurrencias')
```

```python
fig, ax = plt.subplots(figsize=(15, 6))
n_bins = 300

# plot the cumulative histogram
n, bins, patches = ax.hist(
    df['Price'],
    n_bins,
    density=True,
    histtype='step',
    cumulative=True,
    label='Empirical'
    )

# tidy up the figure
ax.grid(True)
ax.legend(loc='right')
ax.set_title('Fig. 3 Función de distribución acumulativa empírica')
ax.set_xlabel('Price de las casas')
ax.set_ylabel('Proporción de ocurrencia')

plt.show()
```
