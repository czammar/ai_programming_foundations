```python
# Cargar el modelo de spaCy para español
# Asegúrate de que el modelo ya ha sido descargado antes de ejecutar este código
nlp = spacy.load("es_core_news_sm")
```

```python
# Introducción a la tokenización de texto
texto = "La tokenización de texto es sencilla y necesaria para el análisis. Ahora lo hacemos con spaCy"

# Procesar el texto con spaCy
doc = nlp(texto)

# Extraer los tokens
tokens = [token.text for token in doc]
print(tokens)
```

```python
# Aplicación del modelo BERT para identificación de sentimientos
from transformers import BertForSequenceClassification, pipeline

# Cargar modelo específico para español
model_name = "finiteautomata/beto-sentiment-analysis"
model = BertForSequenceClassification.from_pretrained(model_name)

# Crear pipeline de análisis de sentimientos
clasificador = pipeline(
    "sentiment-analysis",
    model=model,
    tokenizer=model_name,
    device=-1  # Usar CPU
)

def analizar_ejemplo(texto):
    resultado = clasificador(texto)[0]
    print(f"Texto: {texto}")
    print(f"Sentimiento: {resultado['label']} (Confianza: {resultado['score']:.2%})")
    print("-" * 60)
```

```python
## 1. Tecnología (IA y Software)
print("\n" + "="*20 + " ANÁLISIS DE TECNOLOGÍA " + "="*20)
analizar_ejemplo("Los avances en inteligencia artificial son revolucionarios y prometedores")  # Positivo
analizar_ejemplo("El software tiene bugs frecuentes que afectan su rendimiento")  # Negativo
analizar_ejemplo("La nueva actualización del sistema operativo incluye mejoras menores")  # Neutral

```

```python
## 2. Servicio al Cliente
print("\n" + "="*20 + " ANÁLISIS DE SERVICIO AL CLIENTE " + "="*20)
analizar_ejemplo("El agente de soporte fue extremadamente útil y resolvió mi problema rápido")  # Positivo
analizar_ejemplo("La atención fue pésima, me mantuvieron esperando más de una hora")  # Negativo
analizar_ejemplo("El tiempo de respuesta del servicio técnico fue dentro del promedio")  # Neutral
```

```python
## 3. Productos de Consumo
print("\n" + "="*20 + " ANÁLISIS DE PRODUCTOS " + "="*20)
analizar_ejemplo("Este smartphone superó todas mis expectativas en calidad y desempeño")  # Positivo
analizar_ejemplo("La lavadora se dañó después de un mes de uso, muy mala calidad")  # Negativo
analizar_ejemplo("El televisor tiene buena imagen aunque los altavoces son regulares")  # Neutral
```
