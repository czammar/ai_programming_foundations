## 1. Libraries

```python
import numpy as np
import time
```

## 2. Utilities

```python
def benchmark(func, *args):
    """Utility to time a function's execution."""
    start = time.time()
    func(*args)
    return time.time() - start
```

## 3. Benchmarking

### 3.1 Data creation

```python
# Define a task: squaring elements of an array
data_size = 100_000_000
data = np.random.randint(0, 5000000, data_size)

# Convert data to a list for the loop version
list_data = data.tolist()
```

```python
data[:10]
```

```python
list_data[:10]
```

### 3.2 Implementations with Python and Numpy

```python
# Python loop implementation
def square_with_loop(arr):
    """
    Squared with comprehension list
    """
    return [x**2 for x in arr]

# NumPy vectorized implementation
def square_with_numpy(arr):
    """
    Squared with numpy
    """
    return np.square(arr)
```

### 3.3 Benchmarking

```python
%%time
# Benchmark the loop version
loop_time = benchmark(square_with_loop, list_data)
```

```python
%%time
# Benchmark the vectorized version
vector_time = benchmark(square_with_numpy, data)
```

```python
print(f"Loop version time: {loop_time:.4f} seconds")
print(f"Vectorized version time: {vector_time:.4f} seconds")
print(f"Vectorization Speedup: {loop_time / vector_time:.2f}x")
```
